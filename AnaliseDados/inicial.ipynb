```python
import pandas as pd
import plotly.express as px

dados = pd.read_csv("cancelamentos_sample.csv")

#Tratamento de dados
dados.drop(columns="CustomerID")


display(dados)
```

```python

#Tratamento de dados
display(dados.info())
dados.dropna()
display(dados.info())
```

```python
#Analise de dados ainda

display(dados["cancelou"].value_counts())
display(dados["assinatura"].value_counts())
```

```python
#criando o grafico 
for coluna in dados.columns:
    if coluna != "cancelou":
        grafico = px.histogram(dados,x=coluna,color="cancelou",text_auto=True)

        #exibir o grafico
        grafico.show()
```

```python
#duracao_contrato != mensal
dados = dados[dados["duracao_contrato"]!="Monthly"]

#atraso_pagamento <= 20 dias
dados = dados[dados["dias_atraso"]<=20]

#Ligacao_call_center <=4
dados = dados[dados["ligacoes_callcenter"]<=4]

display(dados["cancelou"].value_counts())
display(dados["cancelou"].value_counts(normalize=True))
```
